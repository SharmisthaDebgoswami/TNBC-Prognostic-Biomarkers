```python
hub = pd.read_csv("Top10_Hub_Genes.csv")

hub
```

```python
genes = hub["Gene"].tolist()

genes
```

```python
survival_results = pd.DataFrame(columns=[
    "Gene",
    "Hazard Ratio",
    "95% CI",
    "Log-rank p-value",
    "Interpretation"
])

survival_results
```

```python
survival_results.loc[len(survival_results)] = [
    "ESR1",
    0.64,
    "0.58–0.71",
    "<1E-16",
    "High expression associated with better overall survival"
]

survival_results
```

```python

survival_results.to_csv(
    "Survival_Analysis_Results.csv",
    index=False
)

print("Survival results saved successfully!")
```

```python
print("Number of genes analysed:", len(survival_results))

print("\nGenes analysed:")
print(survival_results["Gene"].tolist())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.barh(
    survival_results["Gene"],
    survival_results["Hazard Ratio"]
)

plt.axvline(
    x=1,
    linestyle="--",
    linewidth=2,
    color="red"
)

plt.xlabel("Hazard Ratio")
plt.ylabel("Hub Gene")

plt.title(
    "Hazard Ratios of Identified Hub Genes",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    "Figure15_Hazard_Ratios.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```
