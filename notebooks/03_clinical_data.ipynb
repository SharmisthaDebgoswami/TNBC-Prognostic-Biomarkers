# Clinical Data Integration

## Objective

Load and explore the TCGA-BRCA clinical metadata to identify variables that can be used for downstream transcriptomic analysis.

```python
import pandas as pd
from pathlib import Path

PROJECT_DIR = Path.cwd()

RAW_DIR = PROJECT_DIR / "data" / "raw"

clinical_df = pd.read_csv(
    RAW_DIR / "BRCA_clinicalMatrix",
    sep="\t"
)

clinical_df.head()
```

```python
keywords = [
    "er", "pr", "her2", "receptor",
    "subtype", "pam50",
    "tumor", "stage",
    "histology",
    "survival",
    "age",
    "gender",
    "sample"
]

for col in clinical_df.columns:
    col_lower = col.lower()
    if any(keyword in col_lower for keyword in keywords):
        print(col)
```

```python
clinical_complete = clinical_df[
    clinical_df[
        [
            "ER_Status_nature2012",
            "PR_Status_nature2012",
            "HER2_Final_Status_nature2012"
        ]
    ].notna().all(axis=1)
]

print(clinical_complete.shape)
```

```python
tnbc_df = clinical_complete[
    (clinical_complete["ER_Status_nature2012"] == "Negative") &
    (clinical_complete["PR_Status_nature2012"] == "Negative") &
    (clinical_complete["HER2_Final_Status_nature2012"] == "Negative")
]

print("TNBC patients:", len(tnbc_df))
```

```python
non_tnbc_df = clinical_complete.drop(tnbc_df.index)

print("Non-TNBC patients:", len(non_tnbc_df))
```
