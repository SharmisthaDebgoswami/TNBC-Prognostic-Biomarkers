```python
import pandas as pd
from pathlib import Path

PROJECT_DIR = Path.cwd().parent

DATA_DIR = PROJECT_DIR / "data"
RAW_DIR = DATA_DIR / "raw"
PROCESSED_DIR = DATA_DIR / "processed"

print(PROJECT_DIR)
print(RAW_DIR.exists())
print(PROCESSED_DIR.exists())
```

```python
expression_df = pd.read_csv(
    PROCESSED_DIR / "TCGA_BRCA_expression_processed.csv",
    index_col=0
)

print(expression_df.shape)
expression_df.head()
```

```python
clinical_df = pd.read_csv(
    RAW_DIR / "BRCA_clinicalMatrix",
    sep="\t"
)

print(clinical_df.shape)
clinical_df.head()
```

```python
clinical_complete = clinical_df[
    clinical_df[
        [
            "ER_Status_nature2012",
            "PR_Status_nature2012",
            "HER2_Final_Status_nature2012"
        ]
    ].notna().all(axis=1)
]

print(clinical_complete.shape)
```

```python
tnbc_df = clinical_complete[
    (clinical_complete["ER_Status_nature2012"] == "Negative") &
    (clinical_complete["PR_Status_nature2012"] == "Negative") &
    (clinical_complete["HER2_Final_Status_nature2012"] == "Negative")
]

non_tnbc_df = clinical_complete.drop(tnbc_df.index)

print(f"TNBC patients: {len(tnbc_df)}")
print(f"Non-TNBC patients: {len(non_tnbc_df)}")
```

```python
print("Expression sample IDs:")
print(expression_df.columns[:5])

print("\nClinical sample IDs:")
print(clinical_complete["sampleID"].head())
```

```python
common_samples = expression_df.columns.intersection(clinical_complete["sampleID"])

print(f"Matched samples: {len(common_samples)}")
```

```python
expression_matched = expression_df[common_samples]

clinical_matched = (
    clinical_complete
    .set_index("sampleID")
    .loc[common_samples]
)

print(expression_matched.shape)
print(clinical_matched.shape)
```

```python
tnbc_samples = clinical_matched[
    (clinical_matched["ER_Status_nature2012"] == "Negative") &
    (clinical_matched["PR_Status_nature2012"] == "Negative") &
    (clinical_matched["HER2_Final_Status_nature2012"] == "Negative")
].index

non_tnbc_samples = clinical_matched.drop(tnbc_samples).index

expression_tnbc = expression_matched[tnbc_samples]
expression_non_tnbc = expression_matched[non_tnbc_samples]

print(expression_tnbc.shape)
print(expression_non_tnbc.shape)
```

```python
print("Minimum expression value:", expression_matched.min().min())
print("Maximum expression value:", expression_matched.max().max())
print("Mean expression value:", expression_matched.values.mean())
```

```python
gene_means = expression_matched.mean(axis=1)

expression_filtered = expression_matched[gene_means > 1]

print("Original genes:", expression_matched.shape[0])
print("Filtered genes:", expression_filtered.shape[0])
```

```python
expression_tnbc = expression_filtered[tnbc_samples]
expression_non_tnbc = expression_filtered[non_tnbc_samples]

print("TNBC matrix:", expression_tnbc.shape)
print("Non-TNBC matrix:", expression_non_tnbc.shape)
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
```

```python
X = expression_filtered.T

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled.shape)
```

```python
pca = PCA(n_components=2)

principal_components = pca.fit_transform(X_scaled)

print(principal_components.shape)
```

```python
labels = []

for sample in expression_filtered.columns:
    if sample in tnbc_samples:
        labels.append("TNBC")
    else:
        labels.append("Non-TNBC")
```

```python
plt.figure(figsize=(8,6))

for group in ["TNBC", "Non-TNBC"]:
    idx = [i for i, x in enumerate(labels) if x == group]

    plt.scatter(
        principal_components[idx,0],
        principal_components[idx,1],
        label=group,
        alpha=0.7
    )

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA of TCGA Breast Cancer Samples")
plt.legend()

plt.show()
```

```python
expression_filtered.to_csv(
    PROCESSED_DIR / "expression_filtered.csv"
)

clinical_matched.to_csv(
    PROCESSED_DIR / "clinical_matched.csv"
)
```
